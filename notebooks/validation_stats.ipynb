```python
import torch
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
loaded = torch.load("val_scores_epoch_000.pt")
y_true = loaded['y_true']
y_scores = loaded['y_scores']

plt.figure(figsize=(6, 3))

plt.grid(True, which='both', axis='y', linestyle='--', linewidth=0.5)
sns.histplot(x=y_scores, hue=y_true).get_figure()

# Adjust the legend
plt.legend(title='True Label', labels=['Positive', 'Negative'])
plt.xlabel('Predicted Score')

plt.tight_layout()
plt.savefig("val_scores.pdf")
```
